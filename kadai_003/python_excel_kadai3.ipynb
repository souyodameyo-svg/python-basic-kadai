```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill, Font, Border, Side, Alignment

# 1. データの読み込みと連結（Pandas）
df_2022 = pd.read_excel('2022_年間売上表.xlsx', sheet_name='Sheet1')
df_2023 = pd.read_excel('2023_年間売上表.xlsx', sheet_name='Sheet1')
df_concat = pd.concat([df_2022, df_2023], ignore_index=True)

# 2. 商品と売上年ごとに「金額（千円）」を合計
df_result = df_concat.groupby(['商品', '売上年'])[['金額（千円）']].sum().reset_index()

# 3. Excelファイルとして書き出し（一度保存）
file_name = '売上集計表.xlsx'
df_result.to_excel(file_name, index=False) # 画像に合わせてインデックスは非表示

# 4. openpyxlを使用してフォーマットを設定
wb = load_workbook(file_name)
ws = wb.active

# スタイルの定義
header_fill = PatternFill(start_color='F2F2F2', end_color='F2F2F2', fill_type='solid') # 薄いグレー
header_font = Font(bold=True) # 太字
side = Side(style='thin', color='000000') # 細い黒線
border = Border(left=side, right=side, top=side, bottom=side)
alignment = Alignment(horizontal='center', vertical='center')

# ヘッダー行（1行目）へのスタイル適用
for cell in ws[1]:
    cell.fill = header_fill
    cell.font = header_font
    cell.border = border
    cell.alignment = alignment

# データ行への枠線適用
for row in ws.iter_rows(min_row=2, max_row=ws.max_row, min_col=1, max_col=ws.max_column):
    for cell in row:
        cell.border = border

# 保存
wb.save(file_name)

print(f"「{file_name}」を作成しました。画像通りのフォーマットになっています。")
```
